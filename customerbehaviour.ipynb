```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df['Review Rating'].isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age group
labels = ['Young Adult','Adult','Middle Aged','Senior']
df['age_group']= pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
print(df.columns.tolist())
```

```python
#create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Monthly': 30,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df = df.drop('promo_code_used',axis=1)
```

```python

import pandas as pd
from sqlalchemy import create_engine, text

host = "localhost"
port = 5432
database = "customer_behaviour"
user = "postgres"
password = "root"

# Create connection engine
engine = create_engine(
    f"postgresql+psycopg2://{user}:{password}@{host}:{port}/{database}"
)

# Test the connection
with engine.connect() as conn:
    print("PostgreSQL Connection Successful!")
    print(conn.execute(text("SELECT version();")).fetchone())

# Upload our existing preprocessed DataFrame to PostgreSQL as 'customer_shopping'
df.to_sql('customer_shopping', engine, if_exists='replace', index=False)

# Read the table back using pandas sql to verify it works
df_from_db = pd.read_sql("SELECT * FROM customer_shopping LIMIT 100;", engine)
df_from_db.head()
```

```python
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

host = "localhost"
port = 5432
database = "customer_behaviour"
user = "postgres"
password = quote_plus("root")  # the one you set at install

engine = create_engine(
    f"postgresql+psycopg2://{user}:{password}@{host}:{port}/{database}"
)

with engine.connect() as conn:
    print("PostgreSQL Connection Successful!")
    print(conn.execute(text("SELECT version();")).fetchone())

df.to_sql('customer_shopping', engine, if_exists='replace', index=False)

df_from_db = pd.read_sql("SELECT * FROM customer_shopping LIMIT 100;", engine)
df_from_db.head()
```
